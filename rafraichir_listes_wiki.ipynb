# Mettre à jour les listes du wiki XD (véhicules, acteurs)

Ce notebook lit sur **wikixd.fabmob.io** la liste des pages des catégories `Vehicule` et `Acteur`
et crée les fichiers utilisés par le générateur de fiches :
`donnees_wiki/Categorie_Vehicule.txt` et `donnees_wiki/Categorie_Acteur.txt`.

**Mode d'emploi** : *Exécution → Tout exécuter*, puis suivre l'étape 4 pour déposer les fichiers sur GitHub.

## 2. Lire les listes sur le wiki
Ajoute d'autres catégories dans `CATEGORIES` si besoin (ex. `"Evénement"`).

```
from listes_wiki import lire_en_direct, ecrire_fichier

CATEGORIES = ["Vehicule", "Acteur"]

fichiers_crees = []
for cat in CATEGORIES:
    noms = lire_en_direct(cat)
    print(f"Catégorie:{cat} → {len(noms)} noms")
    if noms:
        print("   exemples :", " | ".join(noms[:10]))
        ecrire_fichier(cat, noms)
        fichiers_crees.append(f"donnees_wiki/Categorie_{cat}.txt")
    else:
        print("   ⚠️ le wiki n'a pas répondu : utilise l'étape 3 (copier-coller).")
```

## 3. (Seulement si l'étape 2 a échoué) Copier-coller depuis le navigateur

1. Ouvre https://wikixd.fabmob.io/index.php?title=Catégorie:Vehicule
2. Sélectionne la liste des pages, copie, et colle-la entre les triples guillemets ci-dessous (un nom par ligne ; les lettres seules A, B, C… sont ignorées).
3. S'il y a un lien « page suivante », ouvre-le et colle aussi la suite.
4. Fais de même pour la catégorie Acteur, puis exécute la cellule.

```
COLLE_VEHICULES = """
"""

COLLE_ACTEURS = """
"""

for cat, texte in (("Vehicule", COLLE_VEHICULES), ("Acteur", COLLE_ACTEURS)):
    noms = sorted({l.strip() for l in texte.splitlines()
                   if len(l.strip()) > 1 and not l.strip().startswith("(")})
    if noms:
        ecrire_fichier(cat, noms)
        f = f"donnees_wiki/Categorie_{cat}.txt"
        if f not in fichiers_crees:
            fichiers_crees.append(f)
        print(f"{cat} : {len(noms)} noms enregistrés")
```

## 4. Télécharger les fichiers et les déposer sur GitHub

La cellule ci-dessous télécharge les fichiers sur ton ordinateur. Ensuite :
1. Ouvre https://github.com/GabrielPlassat/generateurfichewiki/tree/main/donnees_wiki
2. *Add file → Upload files*, glisse les fichiers téléchargés, puis *Commit changes*.

L'appli Streamlit se met à jour toute seule (les listes sont relues au plus tard au bout d'une heure ; *Reboot app* pour forcer).

```
from google.colab import files
for f in fichiers_crees:
    files.download(f)
```
